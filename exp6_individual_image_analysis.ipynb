# Individual image layer analysis

This notebook follows one ImageNet image through one DeiT model. It reports the final top-k prediction, the prediction at each residual state, and each attention and FFN update's direct contribution to the final prediction margin. The direct logit attribution uses the final activation's frozen LayerNorm scale, so the component logits reconstruct the model logits.

The comparison class is the final model's strongest alternative. Positive component values move the model toward its final prediction; negative values move it toward the alternative.
